```python
df_pricing = pd.read_csv("future_product_store_dynamic_pricing.csv")
df_pricing.head()
```

```python
conn = sqlite3.connect("retail_pricing.db")
cursor = conn.cursor()
```

```python
df_pricing.to_sql(
    "future_pricing_recommendations",
    conn,
    if_exists="replace",
    index=False
)
```

```python
#price for product in a store
query = """
SELECT date, product_id, store_id, recommended_price
FROM future_pricing_recommendations
WHERE product_id = 'P001'
  AND store_id = 'S001'
ORDER BY date;
"""
pd.read_sql(query, conn)
```

```python
#category wise average price
query = """
SELECT category, AVG(recommended_price) AS avg_price
FROM future_pricing_recommendations
WHERE holiday_promotion = 1
GROUP BY category;
"""
pd.read_sql(query, conn)
```

```python
#Region-wise pricing insight
query = """
SELECT region, AVG(recommended_price) AS avg_price
FROM future_pricing_recommendations
GROUP BY region;
"""
pd.read_sql(query, conn)
```
