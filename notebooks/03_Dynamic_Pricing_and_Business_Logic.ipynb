```python
df = pd.read_csv("cleaned_retail_inventory_data.csv")
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)

df.tail()
```

```python
# Load final versioned model & feature schema
xgb_model = joblib.load("final_xgboost_model_12f.pkl")
feature_cols = joblib.load("final_feature_columns_12f.pkl")

# Safety check
xgb_model.n_features_in_, len(feature_cols)
```

Elasticity

```python
elasticity_df = df[['price', 'units_sold']].copy()
elasticity_df = elasticity_df[(elasticity_df['price'] > 0) & (elasticity_df['units_sold'] > 0)]

elasticity_df['log_price'] = np.log(elasticity_df['price'])
elasticity_df['log_units_sold'] = np.log(elasticity_df['units_sold'])
```

```python
X_elasticity = elasticity_df[['log_price']]
y_elasticity = elasticity_df['log_units_sold']

elasticity_model = LinearRegression()
elasticity_model.fit(X_elasticity, y_elasticity)

price_elasticity = elasticity_model.coef_[0]
price_elasticity
```

Price Elasticity Estimation

A log–log linear regression model is used to estimate the price elasticity of demand. The resulting elasticity coefficient is close to zero, indicating that demand in the given dataset is relatively insensitive to small price changes. This suggests that factors such as inventory availability, competitor pricing, and contextual variables have a stronger influence on demand than price alone.




Data driven computing of category and region , product id and store id


```python
category_stats = df.groupby('category').agg(
    avg_price=('price', 'mean')
)

category_stats['category_price_index'] = (
    category_stats['avg_price'] / df['price'].mean()
)

category_stats
```

```python
# Product-level price behavior
product_stats = df.groupby('product_id').agg(
    avg_price=('price', 'mean')
)

# Relative to overall mean price
product_stats['product_price_index'] = (
    product_stats['avg_price'] / df['price'].mean()
)

product_stats.head()
```

```python
# Store-level price behavior
store_stats = df.groupby('store_id').agg(
    avg_price=('price', 'mean')
)

# Relative to overall mean price
store_stats['store_price_index'] = (
    store_stats['avg_price'] / df['price'].mean()
)

store_stats.head()
```

```python
# Region-level price calibration
region_stats = df.groupby('region').agg(
    avg_price=('price', 'mean')
)

region_stats['region_price_index'] = (
    region_stats['avg_price'] / df['price'].mean()
)

region_stats.head()
```

Adding current date feature for future prediction

```python
today = pd.Timestamp.today().normalize()
future_days = 14

future_dates = pd.date_range(
    start=today + timedelta(days=1),
    periods=future_days
)

future_dates
```

```python
#identyfying holidays
ind_holidays = holidays.India()

future_base = pd.DataFrame({'date': future_dates})
future_base['holiday_promotion'] = future_base['date'].isin(ind_holidays).astype(int)

future_base
```

```python
#hardcoding weather
future_base['weather_condition'] = 'Sunny'
future_base['seasonality'] = future_base['date'].dt.month.map(
    {12:'Winter',1:'Winter',2:'Winter',
     3:'Spring',4:'Spring',5:'Summer',
     6:'Summer',7:'Monsoon',8:'Monsoon',
     9:'Monsoon',10:'Autumn',11:'Autumn'}
)
```

```python
categories = df['category'].unique()
regions = df['region'].unique()
products = df['product_id'].unique()
stores = df['store_id'].unique()

future_rows = []

for _, r in future_base.iterrows():
    for c in categories:
        for reg in regions:
            for p in products:
                for s in stores:
                    future_rows.append({
                        'date': r['date'],
                        'category': c,
                        'region': reg,
                        'product_id': p,
                        'store_id': s,
                        'holiday_promotion': r['holiday_promotion'],
                        'weather_condition': r['weather_condition'],
                        'seasonality': r['seasonality']
                    })

future_df = pd.DataFrame(future_rows)
future_df.head()
```

Price prediction

```python
category_means = df.groupby('category').agg({
    'price': 'mean',
    'competitor_pricing': 'mean',
    'inventory_level': 'mean'
}).reset_index()

future_df = future_df.merge(category_means, on='category', how='left')


```

```python
future_enc = pd.get_dummies(
    future_df,
    columns=['weather_condition', 'seasonality'],
    drop_first=True
)

for col in feature_cols:
    if col not in future_enc.columns:
        future_enc[col] = 0

X_future = future_enc[feature_cols].astype(float)
```

```python
#future demand
future_df['predicted_demand'] = xgb_model.predict(X_future.values)
```

```python
future_df.columns

```

```python
future_df = future_df.merge(
    category_stats[['category_price_index']],
    left_on='category',
    right_index=True,
    how='left'
)
```

```python
future_df = future_df.merge(
    region_stats[['region_price_index']],
    left_on='region',
    right_index=True,
    how='left'
)

```

```python
future_df = future_df.merge(
    product_stats[['product_price_index']],
    left_on='product_id',
    right_index=True,
    how='left'
)

```

```python
future_df = future_df.merge(
    store_stats[['store_price_index']],
    left_on='store_id',
    right_index=True,
    how='left'
)
```

```python
#base price calculation based on split
future_df['calibrated_price'] = (
    future_df['price']
    * future_df['category_price_index']
    * future_df['region_price_index']
    * future_df['product_price_index']
    * future_df['store_price_index']
)

```

```python
#dynamic pricing optimization
def generate_price_range(base_price, steps=7):
    return np.linspace(base_price * 0.9, base_price * 1.1, steps)
```

```python
pricing_rows = []

for _, r in future_df.iterrows():
    for p in generate_price_range(r['calibrated_price']):
        elasticity = r.get('category_elasticity', 1.0)
        adjusted_demand = r['predicted_demand'] * (r['calibrated_price'] / p) ** elasticity
        revenue = p * adjusted_demand

        pricing_rows.append({
            'date': r['date'],
            'category': r['category'],
            'region': r['region'],
            'product_id': r['product_id'],
            'store_id': r['store_id'],
            'predicted_demand': round(r['predicted_demand'], 2),
            'recommended_price': round(p, 2),
            'expected_revenue': round(revenue, 2),
            'holiday_promotion': r['holiday_promotion']
        })
```

```python
final_future_pricing = (
    pricing_df
    .sort_values('expected_revenue', ascending=False)
    .groupby(['date', 'category', 'region', 'product_id', 'store_id'])
    .first()
    .reset_index()
)


```

```python
final_future_pricing.to_csv(
    "future_product_store_dynamic_pricing.csv",
    index=False
)
```
